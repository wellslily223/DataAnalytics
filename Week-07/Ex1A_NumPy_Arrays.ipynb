```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x)

for x in arrA:
    print(x * 2)
```

```python
for y in arrA:
    print(y)

for y in arrA:
   print(y * 2)
```

There's no difference between the outputs themselves - it lists the arrays in the same order. The main difference between the first array and the second array lie in the values - for example, x = 11, 12 and 13 while x * 2 = 22, 24, 26.

## Joining and Stacking Arrays

```python
joined_array_1 = arrA + arrA.sum(axis=0)
print(joined_array_1) 
```

```python
joined_array_2 = arrA.sum(axis=1, keepdims=True) + arrA
print(joined_array_2)
```

```python
stacked_array_1 = joined_array_1 + arrA.sum(axis=0)
print(stacked_array_1)
```

```python
stacked_array_2 = joined_array_2 + arrA.sum(axis=1, keepdims=True)
print(stacked_array_2)
```

```python
another_joined_array = arrA + joined_array_1
print(another_joined_array)
```

```python
another_2 = joined_array_1 + arrA(axis=1)
print(another_2)
```

The type error that comes back describes the numpy.ndarray object as un-callable. 

```python
array1 = np.hstack(arrA)
array2 = np.vstack(arrA)
array3 = np.dstack(arrA)

print(array1)
print(array2)
print(array3)
```

## Searching and Replacing Array Values.

```python
numbers = [36, 39, 42, 39, 42, 45]
result_1 = [n for n in numbers if n % 2 == 0]
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([7, 6, 13, 5, 7, 3])
arrB_sorted = sorted(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
